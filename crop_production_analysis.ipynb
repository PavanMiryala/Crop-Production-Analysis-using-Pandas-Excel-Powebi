```python
# 1. State production
state_production = (
    df.groupby("State_Name")["Production"]
      .sum()
      .sort_values(ascending=False)
)

# 2. Crop production
crop_production = (
    df.groupby("Crop")["Production"]
      .sum()
      .sort_values(ascending=False)
)

# 3. State aggregate productivity
state_productivity = (
    df.groupby("State_Name")
      .agg(
          Total_Area=("Area", "sum"),
          Total_Production=("Production", "sum")
      )
)

state_productivity["Yield"] = (
    state_productivity["Total_Production"] /
    state_productivity["Total_Area"]
)

state_productivity = state_productivity.sort_values(
    "Yield",
    ascending=False
)

# 4. Season production
season_production = (
    df.groupby("Season")["Production"]
      .sum()
      .sort_values(ascending=False)
)

# 5. Crop aggregate productivity
crop_productivity = (
    df.groupby("Crop")
      .agg(
          Total_Area=("Area", "sum"),
          Total_Production=("Production", "sum")
      )
)

crop_productivity["Yield"] = (
    crop_productivity["Total_Production"] /
    crop_productivity["Total_Area"]
)

crop_productivity = crop_productivity.sort_values(
    "Yield",
    ascending=False
)

print("Analysis variables recreated successfully.")
```
